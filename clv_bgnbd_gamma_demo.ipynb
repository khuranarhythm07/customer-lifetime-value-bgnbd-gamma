# Customer Lifetime Value Modeling using BG/NBD and Gamma-Gamma
This notebook demonstrates how to estimate CLV using probabilistic models.

```python
import pandas as pd
from lifetimes.utils import summary_data_from_transaction_data
from lifetimes import BetaGeoFitter, GammaGammaFitter
```

```python
df = pd.read_csv('bgnbd_gamma_model.csv')
df.head()
```

## Generate RFM Metrics

```python
summary = summary_data_from_transaction_data(
    df,
    'customer_id',
    'transaction_date',
    monetary_value_col='monetary_value'
)

summary.head()
```

## Fit BG/NBD Model

```python
bgf = BetaGeoFitter()

bgf.fit(
    summary['frequency'],
    summary['recency'],
    summary['T']
)
```

## Predict Future Purchases

```python
predicted_purchases = bgf.predict(
    12,
    summary['frequency'],
    summary['recency'],
    summary['T']
)

predicted_purchases.head()
```

## Fit Gamma-Gamma Model

```python
ggf = GammaGammaFitter()

ggf.fit(
    summary['frequency'],
    summary['monetary_value']
)
```

## Estimate Customer Lifetime Value

```python
summary['predicted_clv'] = predicted_purchases * \
    ggf.conditional_expected_average_profit(
        summary['frequency'],
        summary['monetary_value']
)

summary.head()
```
